```python
# Find potentially ambiguous numeric dates

date_series = test_scores["date"].astype(str).str.strip()

ambiguous_dates = date_series[
    date_series.str.match(r"^\d{1,2}[-/.]\d{1,2}[-/.]\d{4}$")
].copy()

print("Numeric date records:", len(ambiguous_dates))

print("\nSample numeric dates:")
print(ambiguous_dates.head(30).tolist())

print("\nPotentially ambiguous dates (both first components <= 12):")

def is_ambiguous_date(value):
    parts = value.replace("/", "-").replace(".", "-").split("-")
    
    if len(parts) != 3:
        return False
    
    try:
        first = int(parts[0])
        second = int(parts[1])
        
        # YYYY-MM-DD is not ambiguous
        if len(parts[0]) == 4:
            return False
        
        return first <= 12 and second <= 12
    except:
        return False

ambiguous = ambiguous_dates[ambiguous_dates.apply(is_ambiguous_date)]

print("Ambiguous records:", len(ambiguous))
print(ambiguous.head(50).tolist())
```

```python
# Inspect unambiguous numeric dates to determine the intended format

date_series = test_scores["date"].astype(str).str.strip()

numeric_dates = date_series[
    date_series.str.match(r"^\d{1,2}[-/.]\d{1,2}[-/.]\d{4}$")
]

def classify_date_format(value):
    parts = value.replace("/", "-").replace(".", "-").split("-")
    
    if len(parts) != 3:
        return "Other"
    
    first, second, year = map(int, parts)
    
    if first > 12 and second <= 12:
        return "DD-MM-YYYY"
    elif second > 12 and first <= 12:
        return "MM-DD-YYYY"
    elif first <= 12 and second <= 12:
        return "Ambiguous"
    else:
        return "Invalid"

format_classification = numeric_dates.apply(classify_date_format)

print(format_classification.value_counts())

print("\n--- Clearly DD-MM-YYYY examples ---")
print(numeric_dates[format_classification == "DD-MM-YYYY"].head(30).tolist())

print("\n--- Clearly MM-DD-YYYY examples ---")
print(numeric_dates[format_classification == "MM-DD-YYYY"].head(30).tolist())
```

```python
import re
from datetime import datetime

def clean_test_date(value):
    value = str(value).strip()

    # YYYY-MM-DD / YYYY/MM/DD / YYYY.MM.DD
    if re.match(r"^\d{4}[-/.]\d{1,2}[-/.]\d{1,2}$", value):
        return pd.to_datetime(value, yearfirst=True, errors="coerce")

    # Text month formats: 06-Mar-2026, 10-Apr-2025, etc.
    if re.match(r"^\d{1,2}-[A-Za-z]{3}-\d{4}$", value):
        return pd.to_datetime(value, format="%d-%b-%Y", errors="coerce")

    # Numeric dates
    parts = re.split(r"[-/.]", value)

    if len(parts) == 3:
        first, second, year = parts

        try:
            first = int(first)
            second = int(second)
            year = int(year)

            # Slash or dot → DD/MM/YYYY
            if "/" in value or "." in value:
                return pd.to_datetime(
                    f"{year}-{second:02d}-{first:02d}",
                    errors="coerce"
                )

            # Hyphen
            if "-" in value:
                # If first > 12 → definitely DD-MM-YYYY
                if first > 12:
                    return pd.to_datetime(
                        f"{year}-{second:02d}-{first:02d}",
                        errors="coerce"
                    )

                # If second > 12 → definitely MM-DD-YYYY
                if second > 12:
                    return pd.to_datetime(
                        f"{year}-{first:02d}-{second:02d}",
                        errors="coerce"
                    )

                # Ambiguous hyphen → use MM-DD-YYYY
                return pd.to_datetime(
                    f"{year}-{first:02d}-{second:02d}",
                    errors="coerce"
                )

        except:
            return pd.NaT

    return pd.NaT


# Create cleaned date column
test_scores["date_clean"] = test_scores["date"].apply(clean_test_date)

print("Original dates:", len(test_scores))
print("Successfully parsed:", test_scores["date_clean"].notna().sum())
print("Invalid/unparsed:", test_scores["date_clean"].isna().sum())

print("\nDate range:")
print("Minimum:", test_scores["date_clean"].min())
print("Maximum:", test_scores["date_clean"].max())

print("\nSample conversion:")
display(test_scores[["date", "date_clean"]].head(30))
```

```python
import re

def normalize_school_id(value):
    value = str(value).strip().upper()

    # Extract numeric portion
    numbers = re.findall(r"\d+", value)

    if len(numbers) == 1:
        return f"SCH{int(numbers[0]):04d}"

    return pd.NA

test_scores["school_id_clean"] = (
    test_scores["school_id"]
    .apply(normalize_school_id)
)

print("=== SCHOOL ID CLEANING ===")

print("Original records:", len(test_scores))
print("Missing cleaned IDs:", test_scores["school_id_clean"].isna().sum())
print("Unique raw IDs:", test_scores["school_id"].nunique())
print("Unique cleaned IDs:", test_scores["school_id_clean"].nunique())

print("\nSample original → cleaned:")

display(
    test_scores[["school_id", "school_id_clean"]]
    .drop_duplicates()
    .head(30)
)
```

```python
import numpy as np

# Start with an empty numeric column
test_scores["score_percentage"] = np.nan

# 1. Percentage
mask = test_scores["grading_scale_clean"] == "Percentage"

test_scores.loc[mask, "score_percentage"] = (
    test_scores.loc[mask, "avg_score"]
    .astype(str)
    .str.replace("%", "", regex=False)
    .str.strip()
    .astype(float)
)

# 2. CGPA → Percentage
mask = test_scores["grading_scale_clean"] == "CGPA"

test_scores.loc[mask, "score_percentage"] = (
    pd.to_numeric(
        test_scores.loc[mask, "avg_score"],
        errors="coerce"
    ) * 10
)

# 3. Letter Grade → Percentage
letter_mapping = {
    "A+": 95,
    "A": 85,
    "B": 75,
    "C": 65,
    "D": 55,
    "E": 40
}

mask = test_scores["grading_scale_clean"] == "Letter Grade"

test_scores.loc[mask, "score_percentage"] = (
    test_scores.loc[mask, "avg_score"]
    .astype(str)
    .str.strip()
    .map(letter_mapping)
)

# 4. Raw Marks → Percentage
mask = test_scores["grading_scale_clean"] == "Raw Marks"

raw_parts = test_scores.loc[mask, "avg_score"].astype(str).str.extract(
    r"^\s*([\d.]+)\s*/\s*([\d.]+)\s*$"
)

obtained = pd.to_numeric(raw_parts[0], errors="coerce")
maximum = pd.to_numeric(raw_parts[1], errors="coerce")

test_scores.loc[mask, "score_percentage"] = (
    obtained / maximum * 100
)

print("=== SCORE PERCENTAGE CONVERSION ===")

print("Missing converted scores:",
      test_scores["score_percentage"].isna().sum())

print("Total records:",
      len(test_scores))

print("Minimum:",
      test_scores["score_percentage"].min())

print("Maximum:",
      test_scores["score_percentage"].max())

print("\nAverage score percentage:",
      round(test_scores["score_percentage"].mean(), 2))

print("\nScores by grading scale:")
print(
    test_scores.groupby("grading_scale_clean")["score_percentage"]
    .agg(["count", "min", "max", "mean"])
)

print("\nSample converted values:")
display(
    test_scores[
        ["grading_scale_clean", "avg_score", "score_percentage"]
    ].sample(20, random_state=42)
)
```
